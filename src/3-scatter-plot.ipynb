### Exercise 3: Scatter Plot with Annotations

```python
import matplotlib.pyplot as plt

# Data
x = [1, 2, 3, 4, 5]
y = [2, 4, 6, 8, 10]

# Create scatter plot
plt.scatter(x, y)
plt.title('Scatter Plot with Annotations')
plt.xlabel('X values')
plt.ylabel('Y values')

# Annotate each point
for i in range(len(x)):
    plt.annotate(f'({x[i]}, {y[i]})', (x[i], y[i]))

# Show plot
plt.show()
```
